# TrafficFlowBench 2026 — 01 Exploratory Data Analysis

This notebook explores the freeway corridors, mainline detector observations, masking patterns (R1, R2, R3), and ramp interactions.

```python
import sys
from pathlib import Path

# Add project root to sys.path
ROOT_DIR = Path("..").resolve()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils.data_loader import load_corridor, load_corridor_config, get_corridor_list

print("Environment initialized successfully.")
```

## 1. List Corridors and Inspect Network Configuration

```python
DATA_ROOT = ROOT_DIR / "data" / "kaggle_public"
corridors = get_corridor_list(DATA_ROOT)
print(f"Found {len(corridors)} corridors: {corridors}")

PANEL = "D12_I5_N"
cfg = load_corridor_config(DATA_ROOT, PANEL)
links = cfg.get("links")
if links is not None:
    display(links.head())
    print(f"Total links in {PANEL}: {len(links)}")
```

## 2. Load Sample Mainline Observations

```python
try:
    df, _ = load_corridor(DATA_ROOT, PANEL, split="train", masked=True, sample_n_files=5)
    print(f"Loaded {len(df):,} observations from {PANEL} train split.")
    display(df.head())
except Exception as e:
    print(f"Data not yet unpacked: {e}")
```

## 3. Visualize Fundamental Diagram (Flow vs Speed vs Density)

```python
if 'df' in locals() and not df.empty:
    sample = df.dropna(subset=["speed_kmh", "flow_vph"]).sample(min(10000, len(df)), random_state=42)
    plt.figure(figsize=(10, 5))
    plt.scatter(sample["speed_kmh"], sample["flow_vph"], alpha=0.2, s=8, c="steelblue")
    plt.xlabel("Speed (km/h)")
    plt.ylabel("Flow (vph)")
    plt.title(f"Fundamental Diagram Sample — {PANEL}")
    plt.grid(True, alpha=0.3)
    plt.show()
```
